# ParlaMint-DK Exploration

This notebook investigates the ParlaMint-DK 4.1 dataset and its
CAP-compatible policy-domain annotations.

The goals are to:

1. Inspect the structure and policy annotations of ParlaMint-DK.
2. Determine whether ParlaMint agenda items can be linked to ODA parliamentary cases.
3. Measure how much additional labelled coverage ParlaMint can provide,
   particularly for parliamentary cases after the Danish CAP bills dataset ends.
4. Export any reliable ODA–ParlaMint links for use in the main CAP topic
   classification notebook.

```python
agenda_rows = []

xml_files = list(TEI_DIR.rglob("*.xml"))

for xml_file in xml_files:
    tree = ET.parse(xml_file)
    root = tree.getroot()

    for div in root.findall(".//tei:div[@type='debateSection']", NS):

        head = div.find("tei:head", NS)
        note = div.find("tei:note[@type='agendaItem']", NS)

        if head is None:
            continue

        domains = set()

        for u in div.findall(".//tei:u", NS):
            ana = u.attrib.get("ana", "")

            for token in ana.split():
                if token.startswith("#domain."):
                    domains.add(
                        token.removeprefix("#domain.")
                    )

        agenda_id = (
            "".join(note.itertext()).strip()
            if note is not None
            else None
        )

        agenda_rows.append({
            "source_file": xml_file.name,
            "heading": "".join(head.itertext()).strip(),
            "agenda_id": agenda_id,
            "domains": sorted(domains)
        })

df_parlamint_agenda = pd.DataFrame(agenda_rows)

print("Agenda sections:", len(df_parlamint_agenda))
df_parlamint_agenda.head()
```

```python
def extract_case_number(heading):
    if pd.isna(heading):
        return None

    match = re.search(
        r"\b(L|B|F|R)\s+(\d+)(?:\s*([A-Z]))?\b",
        heading
    )

    if not match:
        return None

    case_type = match.group(1)
    number = match.group(2)
    suffix = match.group(3)

    if suffix:
        return f"{case_type} {number} {suffix}"

    return f"{case_type} {number}"
```

```python
def distance_to_range(row):
    if row["dato"] < row["first_date"]:
        return (row["first_date"] - row["dato"]).days

    if row["dato"] > row["last_date"]:
        return (row["dato"] - row["last_date"]).days

    return 0


candidate_matches["date_distance"] = (
    candidate_matches.apply(distance_to_range, axis=1)
)
```

```python
# Find the minimum temporal distance for each ODA case
min_distances = (
    candidate_matches
    .groupby("sagid")["date_distance"]
    .min()
    .rename("min_date_distance")
)

candidate_matches = candidate_matches.merge(
    min_distances,
    on="sagid",
    how="left"
)

# Keep every candidate tied for best temporal match
best_candidates = candidate_matches[
    candidate_matches["date_distance"]
    == candidate_matches["min_date_distance"]
].copy()

# How many equally-good candidates does each ODA case have?
n_best_candidates = (
    best_candidates
    .groupby("sagid")
    .size()
)

print(n_best_candidates.value_counts().sort_index())
```

```python
import re
from difflib import SequenceMatcher

def normalize_title(text):
    if pd.isna(text):
        return ""

    text = str(text).lower()
    text = re.sub(r"[^\w\s]", " ", text)
    text = re.sub(r"\s+", " ", text).strip()

    return text


def title_similarity(a, b):
    a = normalize_title(a)
    b = normalize_title(b)

    if not a or not b:
        return None

    return SequenceMatcher(None, a, b).ratio()
```

```python
best_matches["title_similarity"] = best_matches.apply(
    lambda row: title_similarity(
        row["sag_titelkort"],
        row["heading_title"]
    ),
    axis=1
)
```
