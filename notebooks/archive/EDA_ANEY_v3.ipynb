Resume does explain a lot better what is actually discussed or what's in the law. 

```python
sag_ids = df_roll_calls["sagid"].dropna().astype(int).unique()

print("Unique sager:", len(sag_ids))
```

```python

sag_ids = set(
    df_roll_calls["sagid"]
    .dropna()
    .astype(int)
    .unique()
)

rows = []
skip = 0
batch_size = 100

while True:
    params = {
        "$select": "id,resume",
        "$top": batch_size,
        "$skip": skip,
    }

    response = requests.get(
        f"{BASE_URL}/Sag",
        params=params
    )
    response.raise_for_status()

    batch = response.json()["value"]

    if not batch:
        break

    # Only keep sager that occur in our roll-call data
    rows.extend(
        row for row in batch
        if row["id"] in sag_ids
    )

    skip += batch_size

df_resumes = pd.DataFrame(rows)
```

```python
has_resume = (
    df_resumes["resume"]
    .fillna("")
    .str.strip()
    .ne("")
)

print(f"Resume coverage: {has_resume.mean():.1%}")
```

```python
df_resumes[has_resume].sample(
    min(20, has_resume.sum()),
    random_state=42
)
```

```python
resume_check = (
    df_roll_calls[
        ["sagid", "sag_nummer", "sag_titel", "sag_titelkort"]
    ]
    .drop_duplicates("sagid")
    .merge(
        df_resumes,
        left_on="sagid",
        right_on="id",
        how="left"
    )
)

resume_check[
    [
        "sag_nummer",
        "sag_titel",
        "sag_titelkort",
        "resume"
    ]
].sample(20, random_state=42)
```

```python
df_roll_calls = df_roll_calls.merge(
    df_resumes[["id", "resume"]],
    left_on="sagid",
    right_on="id",
    how="left"
)

df_roll_calls = df_roll_calls.rename(
    columns={"resume": "sag_resume"}
)

df_roll_calls = df_roll_calls.drop(columns="id")
```
